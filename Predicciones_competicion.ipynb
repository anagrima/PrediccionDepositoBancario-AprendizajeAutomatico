En este cuaderno se carga el modelo final previamente entrenado y almacenado en disco, y se utiliza para generar predicciones sobre el conjunto de datos de competición `bank_competition_13.pkl`. El resultado se guarda en un fichero `predicciones.csv`, siguiendo el formato solicitado en el enunciado.

```python
import pandas as pd
import joblib
```

```python
# cargar modelo final
modelo_final = joblib.load("modelo_final.joblib")
```

```python
# cargar datos de competición
X_comp = pd.read_pickle("bank_competition_13.pkl")

print("Dimensiones del conjunto de competición:", X_comp.shape)
X_comp.head()
```

```python
# cargar datos de competición
X_comp = pd.read_pickle("bank_competition_13.pkl").copy()

# reconstruir la variable creada manualmente en el notebook principal
X_comp["contacted_before"] = (X_comp["pdays"] != -1).astype(int)

# reordenar columnas exactamente como espera el modelo
X_comp = X_comp[modelo_final.feature_names_in_]

# generar predicciones
predicciones = modelo_final.predict(X_comp)

# guardar en CSV
pred_df = pd.DataFrame({"deposit": predicciones})
pred_df.to_csv("predicciones.csv", index=False)

pred_df.head()
```
